```python
from collections import namedtuple
Point = namedtuple('Point', ['x', 'y'])
p = Point(10, 20)
print(p.x, p.y)
```

```python
# deque: Double-ended queue. Perfect for sliding windows in time-series data.
#append :  this will add the element at last index
#appendleft :  this will add the element at first index
#pop :  this will remove the element from last index
#popleft :  this will remove the element from first tindex
from collections import deque
d = deque([1, 2, 3], maxlen=8)
d.append(4) # Removes 1 automatically
d.appendleft(25)
print(d)
```

```python
# Counter: Hashable object counter. Essential for NLP and frequency analysis.
from collections import Counter
words = ["apple", "banana", "apple", "cherry"]
counts = Counter(words)
print(counts.most_common(1))
```

```python
# OrderedDict: Remembers insertion order (standard dict does this now, 
# but OrderedDict has specific methods
# like move_to_end).
from collections import OrderedDict
od = OrderedDict.fromkeys('abcde')
od.move_to_end('b')
```

```python
# defaultdict: Returns a default value for a non-existent key. 
# Prevents KeyError during grouping.
from collections import defaultdict
groups = defaultdict(list)
groups['cluster_a'].append(0.95)
```

```python
# ChainMap: Groups multiple dictionaries into a single view. 
# Useful for managing hyperparameter overrides.

from collections import ChainMap
defaults = {'lr': 0.01, 'batch': 32}
user_configs = {'lr': 0.05}
combined = ChainMap(user_configs, defaults)
```

```python
# UserDict/UserList/UserString: Wrappers to make subclassing built-in types easier.

# The itertools Module
# This module offers functions for creating efficient iterators.
# It is critical for memory management when dealing with large datasets (Big Data).

# count(): Infinite arithmetic progression. Useful for generating unique IDs.

from itertools import count
for i in count(10, 2):
    if i > 14: break
    print(i) # 10, 12, 14
```

```python
# product(): Cartesian product. Best for Grid Search hyperparameter tuning.
```

```python
from itertools import product
lrs = [0.1, 0.01]
epochs = [10, 20]
grid = list(product(lrs, epochs))

print(list(grid))
```

```python
# permutations(): All possible orderings.
# combinations(): All possible subsets (order doesn't matter). 
# Used in feature selection.
```

```python
from itertools import permutations
```

```python
features = ['age', 'income', 'zip']

pair = list(permutations(features, 2))

print(pair)
```

```python
from itertools import combinations
features = ['age', 'income', 'zip']
pairs = list(combinations(features, 2))
print(pairs)
```

```python
# combinations_with_replacement(): Subsets allowing individual elements to repeat.
# accumulate(): Running totals/minimums/maximums. Great for financial time-series.
```

```python
from itertools import accumulate
import operator
data = [1, 2, 3, 4]
running_prod = list(accumulate(data, operator.mul))

print(running_prod)
```

```python
# compress(): Filters data using a boolean mask.

from itertools import compress
data = ['A', 'B', 'C']
mask = [True, False, True]
print(list(compress(data, mask))) # ['A', 'C']
```

```python
# groupby(): Groups consecutive keys. (Note: Data must be sorted first).
```

```python
from itertools import groupby
data = sorted([('A', 1), ('B', 2), ('A', 3)])
for key, group in groupby(data, lambda x: x[0]):
    print(key, list(group))
```

```python
# starmap(): Similar to map(), but used when function arguments are already grouped in tuples.
```

```python
from itertools import starmap
values = [(2, 5), (3, 2)]
print(list(starmap(pow, values))) # [32, 9]
```
