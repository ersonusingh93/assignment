```python
data = [3, 2, 5, 4, 7, 2, 3, 3, 1, 6, 4, 2, 3, 5, 2, 4, 2, 1, 3, 5, 6, 3, 2, 1, 4, 2, 4, 5, 3, 2, 7, 2, 3, 4, 5, 1, 6, 2, 4, 3, 5, 3, 2, 4, 2, 6, 3, 2, 4, 5]

# a. Mean Calculation
mean = sum(data) / len(data)

# b. Median Calculation
sorted_data = sorted(data)
n = len(sorted_data)
if n % 2 == 0:
    median = (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2
else:
    median = sorted_data[n//2]

# c. Mode Calculation
frequency = {}
for x in data:
    frequency[x] = frequency.get(x, 0) + 1
mode = [k for k, v in frequency.items() if v == max(frequency.values())]

print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Mode: {mode}")
```

```python
import statistics

class RentalAnalyzer:
    """Class to analyze car rental duration data."""
    
    def __init__(self, durations):
        self.durations = durations

    def get_mean(self):
        return statistics.mean(self.durations)

    def get_median(self):
        return statistics.median(self.durations)

    def get_mode(self):
        # returns a list in case of multiple modes
        return statistics.multimode(self.durations)

    def generate_report(self):
        """Displays a summary of the analysis."""
        print("--- Car Rental Analysis Report ---")
        print(f"Average Duration (Mean): {self.get_mean():.2f} days")
        print(f"Central Duration (Median): {self.get_median()} days")
        print(f"Most Frequent (Mode): {self.get_mode()} days")
        print("-" * 34)

# Data Initialization
rental_data = [3, 2, 5, 4, 7, 2, 3, 3, 1, 6, 4, 2, 3, 5, 2, 4, 2, 1, 3, 5, 6, 3, 2, 1, 4, 2, 4, 5, 3, 2, 7, 2, 3, 4, 5, 1, 6, 2, 4, 3, 5, 3, 2, 4, 2, 6, 3, 2, 4, 5]

# Execution
analyzer = RentalAnalyzer(rental_data)
analyzer.generate_report()
```

```python
#Question 2 
data = [3, 5, 2, 4, 6, 2, 3, 4, 2, 5, 7, 2, 3, 4, 2, 4, 2, 3, 5, 6, 3, 2, 1, 4, 2, 4, 5, 3, 2, 7, 2, 3, 4, 5, 1, 6, 2, 4, 3, 5, 3, 2, 4, 2, 6, 3, 2, 4, 5, 3]

# a. Range
delivery_range = max(data) - min(data)

# b. Variance
mean = sum(data) / len(data)
# Using n-1 for sample variance
variance = sum((x - mean) ** 2 for x in data) / (len(data) - 1)

# c. Standard Deviation
std_dev = variance ** 0.5

print(f"Range: {delivery_range} days")
print(f"Variance: {variance:.4f}")
print(f"Standard Deviation: {std_dev:.4f} days")
```

```python
import statistics

class ShipmentAnalyzer:
    """A professional tool for analyzing logistics delivery performance."""
    
    def __init__(self, delivery_times):
        self.data = delivery_times

    def calculate_range(self):
        return max(self.data) - min(self.data)

    def calculate_variance(self):
        # statistics.variance uses n-1 (Sample Variance)
        return statistics.variance(self.data)

    def calculate_std_dev(self):
        return statistics.stdev(self.data)

    def display_report(self):
        print("=== Logistics Performance Report ===")
        print(f"Sample Size: {len(self.data)} shipments")
        print(f"Delivery Range: {self.calculate_range()} days")
        print(f"Variance      : {self.calculate_variance():.3f}")
        print(f"Std Deviation : {self.calculate_std_dev():.3f} days")
        print("=" * 36)

# Data Initialization
shipment_data = [3, 5, 2, 4, 6, 2, 3, 4, 2, 5, 7, 2, 3, 4, 2, 4, 2, 3, 5, 6, 3, 2, 1, 4, 2, 4, 5, 3, 2, 7, 2, 3, 4, 5, 1, 6, 2, 4, 3, 5, 3, 2, 4, 2, 6, 3, 2, 4, 5, 3]

# Analysis Execution
analyzer = ShipmentAnalyzer(shipment_data)
analyzer.display_report()
```

```python
import statistics

class ProductPerformance:
    """Analyzes financial performance metrics for a specific product."""
    
    def __init__(self, name, revenue_data):
        self.name = name
        self.revenue = revenue_data

    def get_average(self):
        """Calculates the arithmetic mean."""
        return statistics.mean(self.revenue)

    def get_range(self):
        """Calculates the dispersion (Range)."""
        return max(self.revenue) - min(self.revenue)

    def summary(self):
        """Generates a formatted performance report."""
        avg = self.get_average()
        span = self.get_range()
        
        print(f"--- {self.name} Analysis ---")
        print(f"Mean Revenue : ${avg:.2f}k")
        print(f"Total Range  : ${span}k")
        print(f"Performance Status: {'Stable' if span < (avg * 0.3) else 'Volatile'}")
        print("-" * 25)

# Execution
monthly_data = [120, 150, 110, 135, 125, 140, 130, 155, 115, 145, 135, 130]
analyzer = ProductPerformance("Product Alpha", monthly_data)
analyzer.summary()
```
