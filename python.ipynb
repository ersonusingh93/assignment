```python
def contains_sublist(main_list, sub_list):
    n = len(sub_list)
    # Check every slice of the main list that is the same length as the sublist
    return any(main_list[i : i + n] == sub_list for i in range(len(main_list) - n + 1))

list_a = [1, 2, 3, 4, 5]
list_b = [2, 3, 4]

print(f"Does it contain the sublist? {contains_sublist(list_a, list_b)}")
```

```python
list_of_tuples = [(1, 'A'), (2, 'B'), (3, 'C')]

# The * operator unpacks the list into separate tuple arguments for zip()
numbers, letters = zip(*list_of_tuples)

# Convert them to lists (zip returns tuples by default)
numbers = list(numbers)
letters = list(letters)

print(f"Numbers: {numbers}")
print(f"Letters: {letters}")
```
